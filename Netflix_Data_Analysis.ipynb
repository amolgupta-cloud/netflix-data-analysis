```python
df['type'].value_counts().plot(kind='bar')
plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Number of Titles')
plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)

top_countries.plot(kind='bar')
plt.title('Top 10 Countries by Number of Netflix Titles')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

```python
year_counts = df['release_year'].value_counts().sort_index()

year_counts.plot(kind='line', figsize=(10,5))
plt.title('Netflix Titles Released by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()
```

```python
genre_counts = df['listed_in'].value_counts().head(10)

genre_counts.plot(kind='bar', figsize=(10,5))
plt.title('Top 10 Netflix Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
type_percentage = df['type'].value_counts(normalize=True) * 100
type_percentage
```

```python
type_percentage.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6)
)

plt.title('Movies vs TV Shows Distribution')
plt.ylabel('')
plt.show()
```

```python
rating_counts = df['rating'].value_counts().head(10)

rating_counts.plot(kind='bar', figsize=(10,5))

plt.title('Top 10 Netflix Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

```python
# ==========================================
# NETFLIX DATA ANALYSIS - FINAL ANALYSIS
# ==========================================

# 1. Top 10 Directors
print("TOP 10 DIRECTORS")
print(df['director'].value_counts().head(10))


# 2. Top 10 Directors Chart
director_counts = df['director'].value_counts().head(10)

director_counts.plot(kind='bar', figsize=(10,5))
plt.title('Top 10 Directors on Netflix')
plt.xlabel('Director')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.show()


# 3. Movies vs TV Shows by Year
type_year = pd.crosstab(df['release_year'], df['type'])

type_year.plot(figsize=(10,5))
plt.title('Movies vs TV Shows Released by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()


# 4. Most Recent Release Years
print("TOP 10 RELEASE YEARS")
print(df['release_year'].value_counts().head(10))


# 5. Movie Duration Analysis
movies = df[df['type'] == 'Movie'].copy()

movies['duration_minutes'] = (
    movies['duration']
    .str.extract(r'(\d+)')
    .astype(float)
)

print("MOVIE DURATION STATISTICS")
print(movies['duration_minutes'].describe())


# 6. Average Movie Duration
print("AVERAGE MOVIE DURATION:")
print(round(movies['duration_minutes'].mean(), 2), "minutes")


# 7. Movie Duration Distribution
movies['duration_minutes'].dropna().plot(
    kind='hist',
    bins=20,
    figsize=(10,5)
)

plt.title('Distribution of Movie Durations')
plt.xlabel('Duration (minutes)')
plt.ylabel('Number of Movies')
plt.show()


# 8. Most Common Movie Durations
print("TOP 10 MOVIE DURATIONS")
print(movies['duration_minutes'].value_counts().head(10))


# 9. Individual Genre Analysis
# Split multiple genres into separate rows
genres = df['listed_in'].dropna().str.split(', ').explode()

print("TOP 10 INDIVIDUAL GENRES")
print(genres.value_counts().head(10))


# 10. Individual Genre Chart
genres.value_counts().head(10).plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Top 10 Individual Netflix Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.show()


# 11. Final Dataset Summary
print("===================================")
print("FINAL DATASET SUMMARY")
print("===================================")

print("Total Titles:", len(df))
print("Total Movies:", (df['type'] == 'Movie').sum())
print("Total TV Shows:", (df['type'] == 'TV Show').sum())
print("Total Countries:", df['country'].nunique())
print("Total Directors:", df['director'].nunique())
print("Average Movie Duration:",
      round(movies['duration_minutes'].mean(), 2), "minutes")

print("\nPROJECT ANALYSIS COMPLETE! 🎉")
```
